```python
# calculator.py
a = float(input("Number A: "))
b = float(input("Number B: "))

print("1. Add")
print("2. Subtract")
print("3. Multiply")

option = input("Choose an option: ")

if option == "1":
    print("Result:", a + b)
elif option == "2":
    print("Result:", a - b)
elif option == "3":
    print("Result:", a * b)
else:
    print("Not a valid option")
```
